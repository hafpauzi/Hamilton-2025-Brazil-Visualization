```python
import fastf1
import os

# Define the cache directory
cache_dir = 'f1_cache'

# Create the cache directory if it doesn't exist
if not os.path.exists(cache_dir):
    os.makedirs(cache_dir)

# Enable the cache
fastf1.Cache.enable_cache(cache_dir)
```

```python
#Get HAM lap times for brazil session

#Get all sessions in brazil race
session_brazil = fastf1.get_session(2025, 'Brazil', 'R') # R stands for Race
session_brazil.load()
display(session_brazil)

#assign brazil race laps
session_brazil_laps = session_brazil.laps
display(session_brazil_laps.head()) #get 5 columns of the lap session in brazil

#assign hamilton's brazil race laps
HAM_brazil_laps = session_brazil_laps.loc[session_brazil_laps['Driver'] == 'HAM']
display(HAM_brazil_laps) #display all laps he drove in brazil race
```

```python
#clean columns and remove those not needed
columns_to_drop = ['PitOutTime', 'PitInTime', 'Deleted', 'DeletedReason', 'FastF1Generated', 'IsAccurate', 'Stint']
HAM_brazil_laps_cleaned = HAM_brazil_laps.drop(columns=columns_to_drop)
print("Lewis Hamilton's Brazil lap data after dropping specified columns:")
display(HAM_brazil_laps_cleaned)
```

```python
fastest_lap_index = HAM_brazil_laps_cleaned['LapTime'].idxmin() #get fastest lap based on lap time
slowest_lap_index = HAM_brazil_laps_cleaned['LapTime'].idxmax() #get slowest lap based on lap time

fastest_lap_time_ham = HAM_brazil_laps_cleaned.loc[fastest_lap_index, 'LapTime']
fastest_lap_number_ham = HAM_brazil_laps_cleaned.loc[fastest_lap_index, 'LapNumber']

slowest_lap_time_ham = HAM_brazil_laps_cleaned.loc[slowest_lap_index, 'LapTime']
slowest_lap_number_ham = HAM_brazil_laps_cleaned.loc[slowest_lap_index, 'LapNumber']

print(f"Lewis Hamilton's fastest lap time: {fastest_lap_time_ham} (Lap Number: {int(fastest_lap_number_ham)})")
print(f"Lewis Hamilton's slowest lap time: {slowest_lap_time_ham} (Lap Number: {int(slowest_lap_number_ham)})")
```

```python
fastest_lap = HAM_brazil_laps_cleaned.loc[fastest_lap_index]
slowest_lap = HAM_brazil_laps_cleaned.loc[slowest_lap_index]
# a label-based indexer that allows you to select data by row and column labels.
#  used for accessing a group of rows and columns by label(s) or a boolean array.

fastest_lap_telemetry = fastest_lap.telemetry #.telemetry attribute is specific to the fastf1 library
slowest_lap_telemetry = slowest_lap.telemetry #used to retrieve detailed sensor data for a specific lap.

print("Fastest Lap Telemetry Head:")
display(fastest_lap_telemetry.head())
print("\nSlowest Lap Telemetry Head:")
display(slowest_lap_telemetry.head())
```

The .telemetry attribute is specific to the fastf1 library and is used to retrieve detailed sensor data for a specific lap. When you call .telemetry on a Lap object (which represents a single lap's data), fastf1 fetches a DataFrame containing high-frequency data points for that lap. This includes things like:

* Speed: The speed of the car at various points during the lap.
* RPM: Engine revolutions per minute.
* Throttle: The driver's throttle input
percentage.
* Brake: Whether the brake pedal was pressed.
* nGear: The gear selected.
* X, Y, Z: Spatial coordinates on the track, which we used to plot the circuit layout.

```python
#need to extract the relevant 'X', 'Y', and 'Speed' columns from both the fastest and slowest lap telemetry dataframes
#and then calculate the overall minimum and maximum speed values for consistent color mapping.
fastest_lap_data = fastest_lap_telemetry[['X', 'Y', 'Speed']]
slowest_lap_data = slowest_lap_telemetry[['X', 'Y', 'Speed']]

min_speed = min(fastest_lap_data['Speed'].min(), slowest_lap_data['Speed'].min())
max_speed = max(fastest_lap_data['Speed'].max(), slowest_lap_data['Speed'].max())

print("Fastest Lap Data Head:")
display(fastest_lap_data.head())
print("\nSlowest Lap Data Head:")
display(slowest_lap_data.head())
print(f"\nOverall Minimum Speed: {min_speed}")
print(f"Overall Maximum Speed: {max_speed}")
```

```python
import numpy as np

# Define a rotation angle in degrees. Adjust this value to rotate the circuit.
# Try different angles (e.g., 0, 90, 180, 270, or negative values) to find the best orientation.
rotation_angle_degrees = 90
rotation_angle_radians = np.deg2rad(rotation_angle_degrees)

# Apply rotation to fastest_lap_data
fastest_lap_data_rotated = fastest_lap_data.copy()
fastest_lap_data_rotated['X_rotated'] = fastest_lap_data['X'] * np.cos(rotation_angle_radians) - fastest_lap_data['Y'] * np.sin(rotation_angle_radians)
fastest_lap_data_rotated['Y_rotated'] = fastest_lap_data['X'] * np.sin(rotation_angle_radians) + fastest_lap_data['Y'] * np.cos(rotation_angle_radians)

# Apply rotation to slowest_lap_data
slowest_lap_data_rotated = slowest_lap_data.copy()
slowest_lap_data_rotated['X_rotated'] = slowest_lap_data['X'] * np.cos(rotation_angle_radians) - slowest_lap_data['Y'] * np.sin(rotation_angle_radians)
slowest_lap_data_rotated['Y_rotated'] = slowest_lap_data['X'] * np.sin(rotation_angle_radians) + slowest_lap_data['Y'] * np.cos(rotation_angle_radians)

print(f"Rotated fastest lap data head (rotated by {rotation_angle_degrees} degrees):")
display(fastest_lap_data_rotated.head())
print(f"\nRotated slowest lap data head (rotated by {rotation_angle_degrees} degrees):")
display(slowest_lap_data_rotated.head())
```

```python
import matplotlib.pyplot as plt
import matplotlib.colors as colors

# Create a figure and axes for the plot
fig, ax = plt.subplots(figsize=(10, 10))

# Iterate through the fastest_lap_data_rotated DataFrame, plotting segments of the track
for i in range(len(fastest_lap_data_rotated) - 1):
    ax.plot(
        fastest_lap_data_rotated['X_rotated'].iloc[i:i+2],
        fastest_lap_data_rotated['Y_rotated'].iloc[i:i+2],
        color=plt.cm.plasma(
            colors.Normalize(min_speed, max_speed)(fastest_lap_data_rotated['Speed'].iloc[i])
        ),
        linewidth=3
    )

# Create a ScalarMappable object for the colorbar
sm = plt.cm.ScalarMappable(cmap='plasma', norm=colors.Normalize(min_speed, max_speed))
sm.set_array([])

# Add the colorbar to the plot
cbar = fig.colorbar(sm, ax=ax, orientation='vertical', fraction=0.046, pad=0.04)
cbar.set_label('Speed (km/h)', fontsize=12)

# Set the title of the plot
ax.set_title('Lewis Hamilton Slowest & Fastest - Brazil 2025', fontsize=16)

# Set axis labels
ax.set_xlabel('X coordinate', fontsize=12)
ax.set_ylabel('Y coordinate', fontsize=12)

# Ensure the aspect ratio is equal
ax.set_aspect('equal')

# Remove axis ticks for a cleaner visualization
ax.set_xticks([])
ax.set_yticks([])

# Display the plot
plt.show()
```
